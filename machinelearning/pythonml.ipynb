```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Provide the data (Temperature)
X = np.array([60, 65, 70, 75, 80, 85]).reshape(-1, 1) 

# Ice creams sold on those days
y = np.array([10, 12, 15, 18, 22, 25]) 

# 2. Initialize the model
model = LinearRegression()

# 3. Train the model
model.fit(X, y)

# 4. Predict sales for a 90 degree day
predicted_sales = model.predict([[90]])

print(f"Predicted ice cream sales for a 90°F day: {round(predicted_sales[0])}")
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# 1. The Data
# Features: [Passenger Class (1st, 2nd, or 3rd), Age, IsFemale (1 for yes, 0 for no)]
X = np.array([
    [1, 29, 1],  # 1st class, 29 years old, female
    [3, 2,  0],  # 3rd class, 2 years old, male
    [3, 30, 0],  # 3rd class, 30 years old, male
    [1, 54, 0],  # 1st class, 54 years old, male
    [2, 17, 1],  # 2nd class, 17 years old, female
    [3, 40, 1]   # 3rd class, 40 years old, female
])

# Labels: Did they survive? (1 = Yes, 0 = No)
y = np.array([1, 0, 0, 0, 1, 0])

# 2. Initialize the Logistic Regression model
model = LogisticRegression()

# 3. Train the model
model.fit(X, y)

# 4. Predict for a NEW passenger: 
# Let's test a Jack Dawson scenario: 3rd class, 20 years old, male (0)
jack = [[2, 21, 0]]
prediction = model.predict(jack)
probability = model.predict_proba(jack)

print(f"Would Jack survive? {'Yes' if prediction[0] == 1 else 'No'}")
print(f"Chance of survival: {round(probability[0][1] * 100)}%")
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. The Data (Training text and their labels)
# 1 = Spam, 0 = Real Message (Ham)
emails = [
    "Hey, are we still meeting for lunch today?",
    "CONGRATULATIONS! You won a free cash prize click link now",
    "Dear friend, just checking in to see how you are doing.",
    "URGENT: Your account balance is low. Claim your free bonus.",
    "Can you send me the notes from yesterday's class?"
]
labels = [0, 1, 0, 1, 0]

# 2. Convert Text into Numbers (The "Bag of Words")
# This creates a giant table counting how often every single word appears
vectorizer = CountVectorizer()
X_counts = vectorizer.fit_transform(emails)

# 3. Initialize and Train the "Naïve Bayes" Model
# This model is famous for being incredibly fast at text classification
model = MultinomialNB()
model.fit(X_counts, labels)

# 4. Test it with a brand new, sneaky email
new_email = ["URGENT! Winner of cash prize inside!"]

# We must convert the new email using the exact same vectorizer rules
new_email_counts = vectorizer.transform(new_email)
prediction = model.predict(new_email_counts)

print(f"Is this email Spam? {'YES (Spam)' if prediction[0] == 1 else 'NO (Safe)'}")
```

```python
# A sneaky email trying to bypass the spam filter using a normal word
sneaky_email = ["URGENT free cash prize click here for lunch"]

# Convert it to numbers using our existing dictionary
sneaky_counts = vectorizer.transform(sneaky_email)

# Get the prediction and the raw probability percentages
prediction = model.predict(sneaky_counts)
probabilities = model.predict_proba(sneaky_counts)

print(f"Is this email Spam? {'YES (Spam)' if prediction[0] == 1 else 'NO (Safe)'}")
print(f"Probability it is SAFE: {round(probabilities[0][0] * 100)}%")
print(f"Probability it is SPAM: {round(probabilities[0][1] * 100)}%")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Load your dataset, keeping only the first 2 columns
df = pd.read_csv('spam.csv', encoding='latin-1', usecols=[0, 1])

# Rename the columns
df.columns = ['label_name', 'message']

# ---------- CLEANING ----------
# Deletes empty rows if there are any
df = df.dropna(subset=['message'])

# FIXED: Map text labels ('ham'/'spam') to numbers (0/1)
df['label'] = df['label_name'].map({'ham': 0, 'spam': 1})

X = df['message'].astype(str)
y = df['label']

print(f"Successfully cleaned and loaded {len(df)} rows of data!")
print("\nHere is a preview of the correctly loaded data:")
print(df.head())
# -----------------------------

# 2. Vectorize
vectorizer = CountVectorizer()
X_counts = vectorizer.fit_transform(X)

# 3. Train the Model
model = MultinomialNB()
model.fit(X_counts, y)

# 4. Test it
test_msg = ["You won a cash prize! Call 1-800-SPAM now."]
test_counts = vectorizer.transform(test_msg)
prediction = model.predict(test_counts)

print("\n--- TEST RESULT ---")
print(f"Result: {'SPAM DETECTED' if prediction[0] == 1 else 'MESSAGE SAFE'}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Split the data into Training (80%) and Testing (20%)
# test_size=0.20 means 20% goes to the exam. random_state ensures consistent splitting.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# 2. Vectorize the data correctly
vectorizer = CountVectorizer()

# The model fits and learns the vocabulary ONLY from the training set
X_train_counts = vectorizer.fit_transform(X_train)

# The model ONLY transforms the test set (no learning allowed here!)
X_test_counts = vectorizer.transform(X_test)

# 3. Train the model on the training data
evaluation_model = MultinomialNB()
evaluation_model.fit(X_train_counts, y_train)

# 4. Take the Exam! Make predictions on the hidden test data
predictions = evaluation_model.predict(X_test_counts)

# 5. Calculate the score
# This compares the model's 'predictions' against the real answers 'y_test'
accuracy = accuracy_score(y_test, predictions)

print(f"Total rows used for training: {X_train.shape[0]}")
print(f"Total rows used for testing: {X_test.shape[0]}")
print(f"\nModel Accuracy Score: {round(accuracy * 100, 2)}%")
```

```python
from sklearn.metrics import confusion_matrix

# Generate the confusion matrix grid
cm = confusion_matrix(y_test, predictions)

# Extract the specific counts
# tn = True Negative (Safe emails marked Safe)
# fp = False Positive (Safe emails accidentally marked Spam)
# fn = False Negative (Spam emails accidentally marked Safe)
# tp = True Positive (Spam emails marked Spam)
tn, fp, fn, tp = cm.ravel()

print("--- ERROR BREAKDOWN ---")
print(f"Correctly caught Safe messages (True Negatives): {tn}")
print(f"Correctly caught Spam messages (True Positives): {tp}")
print(f"CRITICAL ERRORS (Safe emails sent to Spam folder): {fp}")
print(f"ANNOYING ERRORS (Spam emails allowed into Inbox): {fn}")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. Load the Customer Churn dataset
df = pd.read_csv('churn_data.csv')

# 2. Data Cleaning & Preprocessing
if 'customerID' in df.columns:
    df = df.drop('customerID', axis=1)

if 'TotalCharges' in df.columns:
    df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
    df = df.dropna(subset=['TotalCharges'])

df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# 3. Handle Text/Categorical Columns (One-Hot Encoding)
df_encoded = pd.get_dummies(df, drop_first=True)

# Separate into Features (X) and Target (y)
X = df_encoded.drop('Churn', axis=1)
y = df_encoded['Churn']

# 4. Train-Test Split (80% training, 20% testing exam)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# 5. Train the Random Forest Model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 6. Test and Evaluate
predictions = model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)

print(f"Random Forest Churn Model Accuracy: {round(accuracy * 100, 2)}%")

# 7. Feature Importance Report
feature_importances = pd.Series(model.feature_importances_, index=X.columns)
print("\nTop 5 Most Important Factors Driving Customer Churn:")
print(feature_importances.nlargest(5))
```

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 1. Create a mock dataset of customer income vs spending score
# (In a real project, you would load this via pd.read_csv('customers.csv'))
data = {
    'AnnualIncome': [15, 16, 17, 18, 19, 80, 85, 90, 95, 100, 20, 88, 12, 92],
    'SpendingScore': [39, 81, 6, 77, 40, 75, 82, 70, 88, 79, 15, 90, 22, 65]
}
df = pd.DataFrame(data)

# 2. Initialize the K-Means model
# n_clusters=3 means we are telling the algorithm: "Find 3 distinct groups in this data"
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

# 3. Fit and predict the clusters
# The model groups every customer into cluster 0, 1, or 2 based on similarities
df['Cluster'] = kmeans.fit_predict(df[['AnnualIncome', 'SpendingScore']])

print("Customer Data with Assigned Clusters:")
print(df)

# 4. Let's see the center points (averages) of each customer group
print("\nCluster Center Coordinates (Income vs Spending):")
print(kmeans.cluster_centers_)
```
